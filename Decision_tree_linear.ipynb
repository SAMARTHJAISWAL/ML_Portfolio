# Hospital cost prediction with a decision tree regressor

Predicts total hospital charges (`TOTCHG`) from age, sex, length of stay and diagnosis group (`APRDRG`). Rows 0-348 are used for training and the rest for testing.

```python
%matplotlib inline
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
patients = pd.read_csv('linear_regression_dataset.csv')
patients = patients.dropna()
```

```python
pt =patients.iloc[:349,:]
test = patients.iloc[349:,:]
```

```python
# Train on the first 349 rows only (pt); the remaining rows are the test set.
# The original code trained on all rows, so the test rows were seen in training.
features = ['AGE', 'FEMALE', 'LOS', 'APRDRG']
X = pt[features]

Y = pt[['TOTCHG']]
```

```python
X_test = test[features]

Y_test = test[['TOTCHG']]
```

```python
X = X.values
Y = Y.values.ravel()
X_test = X_test.values
```

```python
from sklearn.preprocessing import StandardScaler

scalar = StandardScaler()
scalar.fit(X)
scalar2 = StandardScaler()
scalar2.fit(X_test)




scaled_data = scalar.transform(X)
scaled_tdata = scalar2.transform(X_test)

#from sklearn.decomposition import PCA
#pca = PCA(n_components = 2)
#pca2 = PCA(n_components = 2)
#pca.fit(scaled_data)
#pca2.fit(scaled_tdata)
#x_pca = pca.transform(scaled_data)
#t_pca = pca2.transform(scaled_tdata)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
DtReg = DecisionTreeRegressor(ccp_alpha=0.001)
```

```python
DtReg.fit(X, Y)
```

```python
y_pred_t = DtReg.predict((X_test))
```

```python
from sklearn import metrics
```

```python
r_square = metrics.r2_score(Y_test, y_pred_t)
print('R-Square Error associated with Decision Tree Regression is:', r_square)
```

```python
feature_names = features
```

```python
ig = plt.figure(figsize=(25,20))
_ = tree.plot_tree(DtReg,
                   feature_names=feature_names,
                   filled=True,
                   fontsize = 12)
```
